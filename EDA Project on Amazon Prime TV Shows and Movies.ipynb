This project focuses on performing Exploratory Data Analysis (EDA) on the Amazon Prime Movies and TV Shows dataset using Python libraries in Google Colab. Box plots and violin plots were used to analyze runtime distribution across movies and TV shows. Line charts were used to study IMDb score trends across release years for both movies and TV shows. The analysis indicated that TV shows often maintained slightly higher ratings compared to movies over different time periods. Some fluctuations and sudden drops were observed due to missing values or limited records in certain years.

A correlation heatmap was also generated to understand relationships between numerical variables such as IMDb score, TMDb popularity, votes, runtime, and seasons. The heatmap showed moderate positive relationships between IMDb scores and TMDb scores, while most other variables had weak correlations. This indicates that audience ratings are influenced by multiple factors rather than a single attribute.

The insights gained from this project can help OTT platforms like Amazon Prime Video improve their content strategy. Understanding audience preferences, popular genres, and successful runtime patterns can support better investment decisions and personalized recommendations. ## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# check duplicate values
print("Duplicate values in titles dataset:", titles.duplicated().sum())
print("Duplicate values in credits dataset:", credits.duplicated().sum())
````

````python
# remove duplicate values
titles.drop_duplicates(inplace=True)
credits.drop_duplicates(inplace=True)
````

````python
# Verify duplicates removed
print("Duplicates after removal (titles):", titles.duplicated().sum())
print("Duplicates after removal (credits):", credits.duplicated().sum())
````

````python
# Check missing values
print("Missing values in titles dataset:", titles.isnull().sum())
print("Missing values in credits dataset:", credits.isnull().sum())
````

````python
# Handling missing values
titles['description'].fillna('No description available', inplace=True)
titles['age_certification'].fillna('Unknown', inplace=True)
titles['seasons'].fillna(0, inplace=True)
titles['imdb_id'].fillna('No IMDb ID available', inplace=True)
titles['imdb_score'].fillna(0, inplace=True)
titles['imdb_votes'].fillna(0, inplace=True)
titles['tmdb_popularity'].fillna(0, inplace=True)
titles['tmdb_score'].fillna(0, inplace=True)
credits['character'].fillna('Unknown Character', inplace=True)
````

````python
# Check Data type
print("Data types in titles dataset:")
print(titles.dtypes)
print("Data types in credits dataset:")
print(credits.dtypes)
````

````python
# Merge Datasets
dataset = pd.merge(titles, credits, on='id')
````

````python
# Final dataset check
print("Merged dataset shape:", dataset.shape)
print("Merged dataset info:", dataset.info())
print("Merged dataset head:", dataset.head())
````

````python
# Final Duplicates check
print("Final duplicates in merged dataset:", dataset.duplicated().sum())
````

````python
# Final missing values check
print("Final missing values in merged dataset:", dataset.isnull().sum())
````
